```python
# ---
# GCaMP6s Axon Imaging Analysis Notebook
# ICA-based bouton extraction + seed-pixel correlation maps
# ---

import numpy as np
import tifffile
import matplotlib.pyplot as plt
from sklearn.decomposition import FastICA
from sklearn.cluster import AgglomerativeClustering
from scipy.stats import zscore
from scipy.ndimage import gaussian_filter
from skimage.filters import threshold_otsu
from skimage.measure import label, regionprops
import random
import os

# ------------------
# Load image stack
# ------------------
filepath = 'AVG_biT_layer1_hx1_001-1rigidAlign.tif'  # Update if needed
img = tifffile.imread(filepath)  # shape: (T, Y, X)
To, Yo, Xo = img.shape
print(f"Loaded original image stack of shape: {img.shape}")

# Peripheral exclusion
top_margin = 5 # pixels
bottom_margin = 5
side_margin = int(X * 0.05)  # 5% of width (ignore turnaround beam-blanking on image left and right side)

# Build a spatial mask
valid_mask = np.zeros((Yo, Xo), dtype=bool)
valid_mask[top_margin:Yo-bottom_margin, side_margin:Xo-side_margin] = True

# Apply it to all frames
masked_img = img.copy()
masked_img[:, ~valid_mask] = 0
# Tm, Ym, Xm = masked_img.shape
print(f"Loaded edge-masked image stack of shape: {masked_img.shape}")

# Trim off jitter frames
masked_img_trimmed = masked_img[76:]  # now shape = (T_trimmed, Y, X)
T, Y, X = masked_img_trimmed.shape
print(f"Loaded edge-masked, frame-trimmed image stack of shape: {masked_img_trimmed.shape}")

# TODO: PRE-SEGMENT STACK BEFORE CROSS-CORR NEED TO APPLY ICA, etc MASK HERE
    # ICA MASK

data = masked_img_trimmed.reshape(T, -1)
# data_z = zscore(data, axis=0)  # Z-score pixel time series
# Z-score safely across time (ignoring NaNs)
data_z = zscore(data, axis=0, nan_policy='omit')
# After z-scoring, replace any remaining NaNs with 0
data_z = np.nan_to_num(data_z)
```

```python
# ------------------
# ICA Decomposition
# ------------------
n_components = 20
ica = FastICA(n_components=n_components, max_iter=500, random_state=0)
sources = ica.fit_transform(data_z)
maps = ica.mixing_.T.reshape(n_components, Y, X)

# ------------------
# ROI Extraction
# ------------------
roi_masks = []
roi_traces = []
roi_centroids = []

for i in range(n_components):
    spatial_map = gaussian_filter(maps[i], sigma=1)
    try:
        thresh = threshold_otsu(spatial_map)
        binary_mask = spatial_map > thresh
    except ValueError:
        continue

    labeled_mask = label(binary_mask)
    for region in regionprops(labeled_mask):
        if region.area < 5 or region.area > 500:
            continue
        mask = labeled_mask == region.label
        roi_masks.append(mask)
        roi_centroids.append(region.centroid)

        roi_pixels = img[:, mask]
        F = np.mean(roi_pixels, axis=1)
        F0 = np.percentile(F, 10)
        dff = (F - F0) / (F0 + 1e-6)
        roi_traces.append(dff)
```

```python
# ------------------
# Spatial Clustering
# ------------------
if len(roi_centroids) > 1:
    clustering = AgglomerativeClustering(n_clusters=None, distance_threshold=15)
    cluster_labels = clustering.fit_predict(roi_centroids)
else:
    cluster_labels = [0] * len(roi_centroids)
```
