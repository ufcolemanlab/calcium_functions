```python
# ---- ICA ----
n_components = 20
# ica = FastICA(n_components=n_components, max_iter=500, random_state=0)
# sources = ica.fit_transform(data_z)  # temporal signals
# maps = ica.mixing_.T.reshape(n_components, Y, X)  # spatial maps

# 4. ICA on valid, finite data
ica = FastICA(n_components=n_components, max_iter=500, random_state=0)
sources = ica.fit_transform(data_z)
maps = ica.mixing_.T.reshape(n_components, Y, X)  # spatial maps

print('Length of maps var: ' + str(len(maps)))
```

```python
# ---- Segment ROIs from ICA maps ----
roi_masks = []
roi_traces = []

for i in range(n_components):
    spatial_map = gaussian_filter(maps[i], sigma=1)
    try:
        thresh = threshold_otsu(spatial_map)
        binary_mask = spatial_map > thresh
    except ValueError:
        continue  # Skip empty maps

    labeled_mask = label(binary_mask)
    for region in regionprops(labeled_mask):
        if region.area < 10 or region.area > 500:  # skip tiny/noisy blobs
            continue
        roi_mask = labeled_mask == region.label
        roi_masks.append(roi_mask)

        # ---- Extract ∆F/F0 trace ----
        roi_pixels = img[:, roi_mask]
        F = np.mean(roi_pixels, axis=1)
        F0 = np.percentile(F, 10)
        dff = (F - F0) / (F0 + 1e-6)
        roi_traces.append(dff)
```

```python
# ---- Plot Traces and Masks ----
mean_img = np.mean(img, axis=0)
plt.figure(figsize=(8, 6))
plt.imshow(mean_img, cmap='gray')
for i, mask in enumerate(roi_masks):
    yx = np.argwhere(mask)
    y, x = np.mean(yx, axis=0)
    plt.plot(x, y, 'ro')
plt.title("ICA-derived bouton ROIs")
plt.show()

plt.figure(figsize=(10, 6))
for trace in roi_traces:
    plt.plot(trace, alpha=0.6)
plt.title("∆F/F0 traces for ICA-derived boutons")
plt.xlabel("Frame")
plt.ylabel("∆F/F0")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import tifffile
import matplotlib.pyplot as plt
from sklearn.decomposition import FastICA
from scipy.stats import zscore
from skimage.transform import downscale_local_mean

# --- Load 16-bit image stack and optional 8-bit binary mask ---
# ---- Load and preprocess ----
# path = '/Users/jcoleman/UFL Dropbox/Jason Coleman/---UF 2012-2019---/--DATA--/in vivo gcamp analysis/gcamp-callosalAxon fig/axon_data_COOPexample/'
path = '/Users/jcoleman/Documents/--LARGE DATA--/axon_data_COOPexample'
tif_filename = 'AVG_biT_layer1_hx1_001-1rigidAlign.tif'
# mask file generated in FIJI - auto local TH (15 radius, Phansalkar)
mask_filename = 'maskAVG_biT_layer1_hx1_001-1rigidAlign.tif' 

mask_file = os.path.join(
    path, 
    mask_filename
)

tif_file = os.path.join(
    path, 
    tif_filename
)

img = tifffile.imread(tif_file)  # shape: (T, Y, X)
mask = tifffile.imread(mask_file) > 0  # binary mask (True = foreground)

# Optional: downsample both image and mask
downsample_factor = 1  # set to 2 or more to reduce size
if downsample_factor > 1:
    img = downscale_local_mean(img, (1, downsample_factor, downsample_factor))
    mask = downscale_local_mean(mask.astype(np.float32), (downsample_factor, downsample_factor)) > 0.5

T, Y, X = img.shape
masked_img = img * mask[None, :, :]  # apply mask over time

# --- Flatten and filter out flat (static) traces ---
data = masked_img.reshape(T, -1).astype(np.float32)
flat_mask = np.std(data, axis=0) > 0
data_valid = data[:, flat_mask]

# --- Z-score without NaNs ---
data_z = zscore(data_valid, axis=0)
```

```python
data_z
```

```python
# --- ICA ---
n_components = 20
ica = FastICA(n_components=n_components, max_iter=500, random_state=0)
sources = ica.fit_transform(data_z)
mixing = ica.mixing_

# --- Rebuild spatial ICA maps ---
maps = np.zeros((n_components, Y * X), dtype=np.float32)
maps[:, flat_mask] = mixing.T
maps = maps.reshape(n_components, Y, X)

# --- Plot ICA spatial maps ---
for i in range(n_components):
    plt.figure(figsize=(5, 5))
    plt.imshow(maps[i], cmap='inferno')
#     plt.imshow(maps[i], cmap='viridis')
#     plt.imshow(maps[i], cmap='viridis', vmin=0.3, vmax=1)
    plt.title(f'ICA Component {i+1}')
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
# Save plots to PNG files in subfolder

import os
os.makedirs("ica_maps_png", exist_ok=True)

for i in range(n_components):
    plt.figure(figsize=(5, 5))
    plt.imshow(maps[i], cmap='inferno')
    plt.title(f'ICA Component {i+1}')
    plt.axis('off')
    plt.tight_layout()
    plt.savefig(f"ica_maps_png/ica_map_{i+1:02d}.png", dpi=300)
    plt.close()
```
