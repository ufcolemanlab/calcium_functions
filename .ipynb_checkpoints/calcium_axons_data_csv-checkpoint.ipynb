```python
# Import CSV data into Pandas dataframe

import os
import pandas as pd

# 'datadirectory' = folder path string
# path = '/Users/jcoleman/Documents/--LARGE DATA--/#Febo/thy1gcamp6s data set/'
path = '/Users/jcoleman/UFL Dropbox/Jason Coleman/---UF 2012-2019---/--DATA--/in vivo gcamp analysis/gcamp-callosalAxon fig/data/'

# csv_filename = 'mThy6s2_alldrift_D4_001Z1_ROIdata.csv'
csv_filename = 'Results_800.csv' # Note these data points are averaged (e.g., from ~30fps to 5fps ~= 4000frames to 800frames)

csv_file = os.path.join(
    path, 
    csv_filename
)

df = pd.read_csv(csv_file)

print("datafile size (row, column) = " + str(df.shape))
```
